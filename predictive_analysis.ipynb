```python
# Drop duplication
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

handle outliers



```python

column = 'age'
Q1 = df[column].quantile(0.25)
Q3 = df[column].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

```

```python
numeric_cols = ["age", "BMI"]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
```

Handle missing values

```python
#  Drop rows having NA values in 'outcome' column
df = df.dropna(subset=['outcome'])
```

```python
#Fill missing values in the "Respiratory rate" column with the mean of that column
df["Respiratory rate"] = df["Respiratory rate"].fillna(df["Respiratory rate"].mean())
```

```python
#Impute missing values with mean of each column (for the whole dataframe)
df.fillna(df.mean(), inplace=True)
```

Normalization and Scaling

```python
#Normalize the data from 5th column until end
df.iloc[:, 3:] = (df.iloc[:, 3:].apply(lambda x: (x - x.min()) / (x.max() - x.min())))
```

```python
#get value counts for each outcome
df['outcome'].value_counts()
```

Train Test Split

```python

#Separate the features and target variable
X = df.drop(['outcome', 'group' , 'ID'], axis=1)
y = df['outcome']
```

```python
# Split the data into train and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=0)
```

Logistic Regression

```python
#logistic regression model to predict the outcome
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
```

```python
y_pred = logreg.predict(X_test)
```

```python
# Evaluate the model
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
#print these scores
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 score:", f1_score(y_test, y_pred))
```

```python
# SVM model to predict the outcome

from sklearn.svm import SVC
svm = SVC()
svm.fit(X_train, y_train)
# Predict the outcome using the trained model
y_pred = svm.predict(X_test)
# Evaluate thr model

# Print these scores
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 score:", f1_score(y_test, y_pred))
```

```python
streamlit_code = """
import streamlit as st
import joblib
import numpy as np

# Load the trained model
model = joblib.load("random_forest.pkl")

# Original min and max used during training
Age_min = 1
Age_max = 75
BMI_min = 10
BMI_max = 40

def minmax_scale(value, vmin, vmax):
    if vmax == vmin:
        return 0.0
    scaled = (value - vmin) / (vmax - vmin)
    return float(np.clip(scaled, 0.0, 1.0))

st.title("Disease Prediction")
st.write(
    "Enter original values. The app will scale them internally using the same min and max used during training."
)

# Input fields for original values
age_raw = st.number_input(
    "Age in years",
    min_value=float(Age_min),
    max_value=float(Age_max),
    step=1.0,
    value=float((Age_min + Age_max) // 2),
)

gender_input = st.selectbox("Gender", options=["Male", "Female"])

bmi_raw = st.number_input(
    "BMI in kg per m2",
    min_value=float(BMI_min),
    max_value=float(BMI_max),
    step=0.1,
    value=float(round((BMI_min + BMI_max) / 2, 1)),
)

hypertensive_input = st.selectbox("Hypertensive", options=["No", "Yes"])
atrialfibrillation_input = st.selectbox("Atrial Fibrillation", options=["No", "Yes"])

# Convert categorical features to numeric
gender_val = 1.0 if gender_input == "Female" else 0.0
hypertensive_val = 1.0 if hypertensive_input == "Yes" else 0.0
atrialfibrillation_val = 1.0 if atrialfibrillation_input == "Yes" else 0.0

# Scale numeric values
age_scaled = minmax_scale(age_raw, Age_min, Age_max)
bmi_scaled = minmax_scale(bmi_raw, BMI_min, BMI_max)

# Show scaled values for debugging
st.write("Scaled values used for prediction")
st.write(
    {
        "age_scaled": age_scaled,
        "bmi_scaled": bmi_scaled,
        "gender": gender_val,
        "hypertensive": hypertensive_val,
        "atrialfibrillation": atrialfibrillation_val,
    }
)

# Input array in correct order
input_data = np.array(
    [[age_scaled, gender_val, bmi_scaled, hypertensive_val, atrialfibrillation_val]]
)

# Predict
if st.button("Predict Outcome"):
    pred = model.predict(input_data)
    pred_label = int(pred[0])

    # Probability
    try:
        pred_proba = model.predict_proba(input_data)[0]
    except:
        pred_proba = None

    outcome_text = "Positive (Disease)" if pred_label == 1 else "Negative (No Disease)"
    st.success(f"Predicted Outcome: {outcome_text}")

    if pred_proba is not None:
        st.write(f"Prediction Probability (Negative, Positive): {pred_proba}")
    else:
        st.write("Model does not support probability output")

"""
```

```python
%%writefile -a app.py
numeric_cols = ["age", "BMI"]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
```

**Reasoning**:
The next step is to append the code to drop rows with missing values in the 'outcome' column to the `app.py` file, as per the instructions.



```python
%%writefile -a app.py
df = df.dropna(subset=['outcome'])
```

**Reasoning**:
The next step is to append the code to fill missing values in the 'Respiratory rate' column with its mean to the `app.py` file, as per the instructions.



```python
%%writefile -a app.py
df["Respiratory rate"] = df["Respiratory rate"].fillna(df["Respiratory rate"].mean())
```

**Reasoning**:
The next step is to append the code to fill any remaining missing values in the entire DataFrame with the mean of their respective columns to the `app.py` file, as per the instructions.



```python
%%writefile -a app.py
df.fillna(df.mean(numeric_only=True), inplace=True)
```

**Reasoning**:
The next step is to append the code to normalize the data from the 4th column (index 3) until the end of the DataFrame using Min-Max Scaling, as specified in the instructions. This completes the data preprocessing steps.



```python
%%writefile -a app.py
df.iloc[:, 3:] = (df.iloc[:, 3:].apply(lambda x: (x - x.min()) / (x.max() - x.min())))
```
